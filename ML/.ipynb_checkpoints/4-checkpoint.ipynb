```python
import matplotlib.pyplot as plt
```

```python
def f(x):
    return (x + 3) ** 2

def df(x):
    return 2 * (x + 3)

x = 2
learning_rate = 0.1
iterations = 20

values = []

for i in range(iterations):
    x = x - learning_rate * df(x)
    values.append(f(x))

print(f"\nMinimum value of x: {x:.2f}")
print(f"\nMinimum value of y: {f(x):.2f}")

plt.plot(values, marker='o')
plt.xlabel("Iterations")
plt.ylabel("Function value")
plt.title("Gradient Descent")
plt.show()
    
```
